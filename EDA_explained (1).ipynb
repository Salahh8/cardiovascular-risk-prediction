```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('ggplot')
import seaborn as sns
```

```python
df = pd.read_csv("train.csv")
```

Loads the libraries needed for data handling and visualization.

 We are preparing the notebook to explore the dataset and create graphs.


```python
fig, ax = plt.subplots(figsize=(15,6))
age_dis=pd.DataFrame(df.groupby(['age'])['id'].count())
sns.barplot(x=age_dis.index,y=age_dis['id'])
plt.ylabel('Counts')
plt.title('Age Distrubution')
```

Loads `train.csv` into a DataFrame called `df`.

 This is the dataset we will analyze.


```python
plt.rcParams['figure.figsize'] = (15, 5)
df.groupby(['age','TenYearCHD'])['id'].count().unstack().plot(kind='bar')
plt.title('Age wise Effected People')
```

The number of people at each age.

 The dataset contains different age groups, with some ages having more participants than others.


```python
plt.rcParams['figure.figsize'] = (20, 5)
df.groupby(['age','sex'])['TenYearCHD'].count().unstack().plot(kind='bar')
```

Each age group is split by `TenYearCHD` (0 = no CHD, 1 = CHD).

 The number of CHD cases is smaller than the number of non-CHD cases, and CHD becomes more noticeable in older age groups.


## Education


```python
fig, ax = plt.subplots(1,2,figsize=(15,5))
ax1=plt.subplot(1,2,1)
Education_status=pd.DataFrame(df.groupby(['education'])['id'].count())
sns.barplot(x=Education_status.index,y=Education_status['id'])
plt.xlabel('Education')
plt.ylabel('Counts')
plt.title('season Distribution in year')
ax2=plt.subplot(1,2,2)
df.groupby(['education'])['id'].count().plot(kind='pie')
plt.title('Education Proposanate')
```

```python
plt.rcParams['figure.figsize'] = (10, 5)
df.groupby(['education','TenYearCHD'])['id'].count().unstack().plot(kind='bar')
plt.title('Education with Ten years CHD')
```

Education levels compared with `TenYearCHD`.

 Most observations are in some education categories more than others. The graph does not by itself show that education causes CHD.
